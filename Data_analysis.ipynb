```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("series \n", series)
```

```python
#Creating a series from a dictionary

data = {"a" : 1 , "b" : 2 , "c" : 3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ["a" , "b" , "c"]
pd.Series(data , index)
```

```python
##Dataframe creation
data = {
    'Name' : ['Siddharth' , 'Nikhil' , 'Titli'],
    'Age' : [21 , 21, 20],
    'city' : ['jaipur' , "sikar" , "jaipur"]

}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
arr = np.array(df)
print(arr)
```

```python
## creating a dataframe form a list of dictionaries

data = [
    {'Name' : 'Siddharth', 'Age' : 21 , "City" : "Jaipur"},
    {'Name' : 'Nikhil', 'Age' : 21 , "City" : "Sikar"},
    {'Name' : 'Sidd', 'Age' : 24 , "City" : "Jodhpur"},
    {'Name' : 'Titli', 'Age' : 20 , "City" : "Jaipur"},
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
