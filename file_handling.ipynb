# Python File Handling Tutorial

File handling is an essential part of any application. Python provides built-in functions and modules to interact with files on your system. In this notebook, we'll cover basic file operations.

## 1. Creating and Writing to a File

```python
# To write to a file, open it in 'w' mode (write mode).
# If the file doesn't exist, it will be created. If it does exist, its content will be truncated (erased).
file_name = 'my_first_file.txt'

f = open(file_name, 'w')
f.write('Hello, this is the first line.\n')
f.write('This is the second line.\n')
f.close()

print(f"File '{file_name}' created and written to.")
```

## 2. Reading from a File

```python
# To read from a file, open it in 'r' mode (read mode).
# It's good practice to close the file after you're done with it.
file_name = 'my_first_file.txt'

f = open(file_name, 'r')
content = f.read()
f.close()

print(f"Content of '{file_name}':\n{content}")
```

## 3. Appending to a File

```python
# To add content to an existing file without overwriting, open it in 'a' mode (append mode).
file_name = 'my_first_file.txt'

f = open(file_name, 'a')
f.write('This line was appended.\n')
f.close()

print(f"Content appended to '{file_name}'.")
```

```python
# Let's read the file again to see the appended content.
file_name = 'my_first_file.txt'

f = open(file_name, 'r')
updated_content = f.read()
f.close()

print(f"Updated content of '{file_name}':\n{updated_content}")
```

## 4. Using the `with` Statement (Recommended)

The `with` statement ensures that the file is automatically closed after its block is exited, even if errors occur. This is the recommended way to handle files in Python.

```python
# Writing with 'with'
file_name_with = 'my_second_file.txt'

with open(file_name_with, 'w') as f:
    f.write('This file was created using the `with` statement.\n')
    f.write('It automatically closes the file.\n')

print(f"File '{file_name_with}' created and written using `with`.")

# Reading with 'with'
with open(file_name_with, 'r') as f:
    content_with = f.read()

print(f"Content of '{file_name_with}' read using `with`:\n{content_with}")
```

## 5. Error Handling: `FileNotFoundError`

When trying to open a file that doesn't exist in read ('r') or append ('a') modes, Python will raise a `FileNotFoundError`. It's good practice to handle such exceptions using `try-except` blocks.

```python
non_existent_file = 'non_existent_file.txt'

try:
    with open(non_existent_file, 'r') as f:
        content = f.read()
        print(content)
except FileNotFoundError:
    print(f"Error: The file '{non_existent_file}' was not found.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

## 6. Deleting a File

To delete a file, you can use the `os` module, which provides functions for interacting with the operating system.

```python
import os

file_to_delete = 'my_first_file.txt'

if os.path.exists(file_to_delete):
    os.remove(file_to_delete)
    print(f"File '{file_to_delete}' deleted successfully.")
else:
    print(f"File '{file_to_delete}' does not exist.")

file_to_delete_with = 'my_second_file.txt'
if os.path.exists(file_to_delete_with):
    os.remove(file_to_delete_with)
    print(f"File '{file_to_delete_with}' deleted successfully.")
else:
    print(f"File '{file_to_delete_with}' does not exist.")
```
