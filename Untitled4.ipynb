# Comprehensive NumPy Notebook

NumPy (Numerical Python) is a fundamental package for scientific computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a collection of high-level mathematical functions to operate on these arrays.

## 1. Importing NumPy

```python
import numpy as np
print("NumPy imported successfully!")
```

## 2. Array Creation

NumPy arrays are the core of the library. They are similar to Python lists but offer much better performance and functionality for numerical operations. Let's explore various ways to create them.

### 2.1 From Python Lists or Tuples

```python
# 1-D array from a list
list_data = [1, 2, 3, 4, 5]
array_1d = np.array(list_data)
print("1-D Array:", array_1d)
print("Type:", type(array_1d))

# 2-D array from a list of lists
list_of_lists = [[1, 2, 3], [4, 5, 6]]
array_2d = np.array(list_of_lists)
print("\n2-D Array:\n", array_2d)

# Array from a tuple
tuple_data = (10, 20, 30)
array_from_tuple = np.array(tuple_data)
print("\nArray from tuple:", array_from_tuple)
```

### 2.2 Using Built-in NumPy Functions

NumPy provides functions to create arrays with initial placeholder content, such as zeros, ones, or random values.

```python
# Array of zeros
zeros_array = np.zeros((3, 4))
print("Array of Zeros (3x4):\n", zeros_array)

# Array of ones
ones_array = np.ones((2, 3), dtype=int) # Specify data type
print("\nArray of Ones (2x3, int):\n", ones_array)

# Array with a constant value
full_array = np.full((2, 2), 7.0)
print("\nArray with Constant Value (2x2, value 7.0):\n", full_array)

# Empty array (content is random, depends on memory)
empty_array = np.empty((2, 2))
print("\nEmpty Array (2x2):\n", empty_array)
```

### 2.3 Generating Sequences

```python
# `arange` (like Python's `range`)
range_array = np.arange(0, 10, 2) # start, stop (exclusive), step
print("arange(0, 10, 2):", range_array)

# `linspace` (linearly spaced numbers)
linspace_array = np.linspace(0, 10, 5) # start, stop (inclusive), number of elements
print("\nlinspace(0, 10, 5):", linspace_array)

# `logspace` (logarithmically spaced numbers)
logspace_array = np.logspace(0, 2, 5) # 10^0 to 10^2, 5 elements
print("\nlogspace(0, 2, 5):", logspace_array)
```

### 2.4 Identity Matrix and Diagonal Arrays

```python
# Identity matrix
identity_matrix = np.eye(3) # Creates a 3x3 identity matrix
print("Identity Matrix (3x3):\n", identity_matrix)

# Diagonal matrix from a 1D array
diag_array = np.diag([1, 2, 3])
print("\nDiagonal Matrix from [1,2,3]:\n", diag_array)
```

### 2.5 Data Types (dtype)

NumPy allows explicit control over the data type of the elements in an array, which is crucial for memory efficiency and numerical precision.

```python
int_array = np.array([1, 2, 3], dtype=np.int16)
print("Integer Array (int16):", int_array, "\tDtype:", int_array.dtype)

float_array = np.array([1.0, 2.5, 3.7], dtype=np.float64)
print("Float Array (float64):", float_array, "\tDtype:", float_array.dtype)

bool_array = np.array([True, False, True], dtype=bool)
print("Boolean Array (bool):", bool_array, "\tDtype:", bool_array.dtype)

# You can also change dtype after creation
changed_dtype = array_1d.astype(np.float32)
print("\nOriginal array_1d (int64):", array_1d.dtype)
print("Changed to float32:", changed_dtype, "\tDtype:", changed_dtype.dtype)
```
