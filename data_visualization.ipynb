# Interactive Data Visualization with Matplotlib

This notebook will guide you through creating interactive data visualizations using `matplotlib`, a popular Python plotting library.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Let's generate some sample data to visualize.

```python
# Generate sample data
np.random.seed(42)
x = np.linspace(0, 10, 100)
y1 = np.sin(x) + np.random.normal(0, 0.1, 100)
y2 = np.cos(x) + np.random.normal(0, 0.1, 100)

df_sample = pd.DataFrame({'x': x, 'y1': y1, 'y2': y2})
```

Now, let's create a simple line plot of the generated data.

```python
plt.figure(figsize=(10, 6))
plt.plot(df_sample['x'], df_sample['y1'], label='Sine Wave')
plt.plot(df_sample['x'], df_sample['y2'], label='Cosine Wave')
plt.title('Sine and Cosine Waves with Noise')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```
