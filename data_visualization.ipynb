# Interactive Data Visualization with Matplotlib

This notebook will guide you through creating interactive data visualizations using `matplotlib`, a popular Python plotting library.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Let's generate some sample data to visualize.

```python
# Generate sample data
np.random.seed(42)
x = np.linspace(0, 10, 100)
y1 = np.sin(x) + np.random.normal(0, 0.1, 100)
y2 = np.cos(x) + np.random.normal(0, 0.1, 100)

df_sample = pd.DataFrame({'x': x, 'y1': y1, 'y2': y2})
```

Now, let's create a simple line plot of the generated data.

```python
plt.figure(figsize=(10, 6))
plt.plot(df_sample['x'], df_sample['y1'], label='Sine Wave')
plt.plot(df_sample['x'], df_sample['y2'], label='Cosine Wave')
plt.title('Sine and Cosine Waves with Noise')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

### Adding Interactivity with `ipywidgets`

To make our plot interactive, we can use `ipywidgets` to control parameters of our data visualization. Let's add a slider to control the amplitude of the sine wave.

```python
from ipywidgets import interact, FloatSlider
from IPython.display import display

def plot_interactive_sine(amplitude=1.0):
    plt.figure(figsize=(10, 6))

    # Recalculate y1 with the new amplitude
    y1_interactive = amplitude * np.sin(df_sample['x']) + np.random.normal(0, 0.1, len(df_sample['x']))

    plt.plot(df_sample['x'], y1_interactive, label=f'Sine Wave (Amplitude: {amplitude:.1f})')
    plt.plot(df_sample['x'], df_sample['y2'], label='Cosine Wave')
    plt.title('Interactive Sine and Cosine Waves with Noise')
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.legend()
    plt.grid(True)
    plt.show()

# Create an interactive slider for amplitude
amplitude_slider = FloatSlider(
    value=1.0,
    min=0.1,
    max=2.0,
    step=0.1,
    description='Sine Amplitude:',
    continuous_update=True
)

# Display the interactive plot
interact(plot_interactive_sine, amplitude=amplitude_slider);
```
